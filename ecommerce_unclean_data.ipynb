```python
# Feature Engineering

# 1. Total Order Amount
df_clean["Total_Amount"] = (
    df_clean["Qty"] * df_clean["Unit_Price"]
)

# 2. Discount Amount
df_clean["Discount_Amount"] = (
    df_clean["Total_Amount"] * df_clean["Discount"] / 100
)

# 3. Final Amount after Discount
df_clean["Final_Amount"] = (
    df_clean["Total_Amount"] - df_clean["Discount_Amount"]
)

# 4. Order Year
df_clean["Order_Year"] = df_clean["Order_Date"].dt.year

# 5. Order Month
df_clean["Order_Month"] = df_clean["Order_Date"].dt.month

# 6. Order Month Name
df_clean["Order_Month_Name"] = (
    df_clean["Order_Date"].dt.month_name()
)

# 7. Order Day
df_clean["Order_Day"] = df_clean["Order_Date"].dt.day

# 8. Order Day Name
df_clean["Order_Day_Name"] = (
    df_clean["Order_Date"].dt.day_name()
)

# 9. Weekend Indicator
df_clean["Is_Weekend"] = (
    df_clean["Order_Date"].dt.dayofweek >= 5
)

# 10. Delivery Days
df_clean["Delivery_Days"] = (
    df_clean["Delivery_Date"] - df_clean["Order_Date"]
).dt.days

# 11. High Value Order
df_clean["High_Value_Order"] = (
    df_clean["Final_Amount"] >= 5000
)

# 12. Quantity Category
df_clean["Quantity_Category"] = pd.cut(
    df_clean["Qty"],
    bins=[0, 2, 4, 5],
    labels=["Low", "Medium", "High"]
)

# Check new features
df_clean.head()
```
